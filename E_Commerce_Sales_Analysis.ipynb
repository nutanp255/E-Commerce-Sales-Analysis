## 🧹 Data cleaning process

```python
#check missing values in each column
df.isna().sum()
```

```python
#Treat missing values
object_value=[]
numeric_value=[]
object_value=df.select_dtypes(include='object').columns.tolist()
numeric_value=df.select_dtypes(include='number').columns.tolist()
object_value
```

```python
numeric_value
```

```python
print("Total Missing Values:", df.isnull().sum().sum())

print("Completely Empty Rows:", df.isnull().all(axis=1).sum())

print("Total Rows:", len(df))
```

```python
#fill numerical columns missing values
for col in numeric_value:
    df[col] = df[col].fillna(df[col].mean())
```

```python
#fill categorical columns missing values
for col in object_value:
    if df[col].isnull().sum() > 0:
        if not df[col].mode().empty:
            df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#check missing values filled or not
df.isna().sum()
```

## 📊 Data Visualization

Data visualization is the graphical representation of data using charts and graphs. In this project, Matplotlib and Seaborn are used to visualize sales, profit, discounts, and product trends. It helps identify patterns, compare performance, and understand business insights easily.


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### 📊 Que.1:	What are the total sales, profit, and quantity?

```python

total = df[['Sales', 'Profit', 'Quantity']].sum()
total.plot(kind='bar')
plt.title('Total Sales, Profit and Quantity')
plt.ylabel('Total')
plt.show()
```

**Analysis:**
Sales is the highest, followed by Profit, while Quantity has the lowest value in the bar chart.

**Conclusion:**
The business generates high sales compared to profit. Further analysis can help identify ways to improve profitability

### 🏆 Que.2:	What are the top 10 products by sales?

```python
top_10_products = df.groupby('Product Name')['Sales'].sum().sort_values(ascending=False).head(10)

top_10_products.sort_values().plot(kind='barh', figsize=(10, 6))

plt.title('Top 10 Products by Sales')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.show()
```

**Analysis:**
The horizontal bar chart shows the top 10 products with the highest sales. It helps identify the best-selling products.

**Conclusion:**
The products with the highest sales contribute significantly to total revenue. The business can focus on these products to improve sales performance

### 🛍️ Que.3:	What is the sales share of each category?

```python
category_sales = df.groupby('Category')['Sales'].sum()

plt.figure(figsize=(7, 7))
plt.pie(category_sales, labels=category_sales.index,
        autopct='%1.1f%%', startangle=90)

plt.title('Sales Share of Each Category')
plt.show()
```

**Analysis:**
Technology has the highest sales share, followed by Furniture, Office Supplies, and Clothing.

**Conclusion:**
Technology contributes the most to total sales. The business should focus on improving sales in the lower-performing categories

### 💸 Que.4:	How are discounts related to profit?

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Discount'], df['Profit'])

plt.title('Discount vs Profit')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.show()
```

**Analysis:**
The scatter plot shows a negative relationship between discounts and profit. As discounts increase, profit generally decreases.

**Conclusion:**
High discounts may lead to losses. The business should control discount levels to maintain profitability

### 📉 Que.5:	How does profit distribution vary?

```python
plt.figure(figsize=(8, 5))

plt.hist(df['Profit'], bins=20, edgecolor='black')

plt.title('Profit Distribution')
plt.xlabel('Profit')
plt.ylabel('Count')

plt.show()
```

**Analysis:**
The histogram shows that most transactions have low to moderate profits. Some transactions result in losses, while a few generate high profits.

**Conclusion:**
The business should focus on reducing losses and increasing profitability through better sales and discount strategies

### 🔗 Que.6:Que.6: What is the correlation between numerical variables?

```python
numeric_df = df.select_dtypes(include='number')

correlation = numeric_df.corr()

plt.figure(figsize=(10, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Between Numerical Variables')
plt.show()
```

**Analysis**
The heatmap shows the correlation between numerical variables. Values close to +1 indicate a positive relationship, while values close to -1 indicate a negative relationship.

**Conclusion**
Correlation helps identify relationships between numerical variables and supports better data-driven business decisions.
